## Виконання

```python
import heapq
```

### Створення вузла дерева Гафмена

```python
# Клас вузла дерева Гафмена
class Node:
    def __init__(self, freq, symbol, left=None, right=None):
        self.freq = freq
        self.symbol = symbol
        self.left = left
        self.right = right
        self.huff = ''

    # Перевантаження оператора <
    def __lt__(self, other):
        return self.freq < other.freq
```

### Функція виведення кодів Гафмена

```python
# Рекурсивна функція для виведення кодів
def print_nodes(node, value=''):
    new_value = value + str(node.huff)

    if node.left:
        print_nodes(node.left, new_value)

    if node.right:
        print_nodes(node.right, new_value)

    if not node.left and not node.right:
        print(f'{node.symbol} -> {new_value}')
```

### Вхідні дані

```python
# Символи
chars = ['a', 'b', 'c', 'd', 'e', 'f']

# Частоти символів
freq = [5, 9, 12, 13, 16, 45]

# Створення списку вузлів
nodes = []

# Додавання вузлів у купу
for i in range(len(chars)):
    heapq.heappush(nodes, Node(freq[i], chars[i]))
```

### Побудова дерева Гафмена

```python
# Побудова дерева
while len(nodes) > 1:

    # Отримуємо два вузли з мінімальною частотою
    left = heapq.heappop(nodes)
    right = heapq.heappop(nodes)

    # Призначаємо коди
    left.huff = 0
    right.huff = 1

    # Створення нового вузла
    new_node = Node(
        left.freq + right.freq,
        left.symbol + right.symbol,
        left,
        right
    )

    # Додаємо новий вузол назад у купу
    heapq.heappush(nodes, new_node)
```

### Виведення кодів Гафмена

```python
print("Коди Гафмена:\n")

print_nodes(nodes[0])
```

### Процедура автоматичного підрахунку частот символів

```python
# Функція для підрахунку частот символів
def calculate_frequency(text):

    frequency = {}

    for char in text:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    chars = list(frequency.keys())
    freq = list(frequency.values())

    return chars, freq
```

### Перевірка роботи функції

```python
text = "huffman coding example"

chars, freq = calculate_frequency(text)

print("Символи:")
print(chars)

print("\nЧастоти:")
print(freq)
```

### Кодування повідомлення

```python
# Словник кодів
codes = {}

# Функція створення словника кодів
def create_codes(node, value=''):

    new_value = value + str(node.huff)

    if node.left:
        create_codes(node.left, new_value)

    if node.right:
        create_codes(node.right, new_value)

    if not node.left and not node.right:
        codes[node.symbol] = new_value
```

```python
# Створення кодів
create_codes(nodes[0])

print(codes)
```

```python
# Функція кодування повідомлення
def encode(text, codes):

    encoded_text = ''

    for char in text:
        encoded_text += codes[char]

    return encoded_text
```

```python
message = "face"

encoded = encode(message, codes)

print("Закодоване повідомлення:")
print(encoded)
```

### Декодування повідомлення

```python
# Функція декодування
def decode(encoded_text, root):

    decoded_text = ''
    current = root

    for bit in encoded_text:

        if bit == '0':
            current = current.left
        else:
            current = current.right

        # Якщо дійшли до листка
        if not current.left and not current.right:
            decoded_text += current.symbol
            current = root

    return decoded_text
```

```python
decoded = decode(encoded, nodes[0])

print("Декодоване повідомлення:")
print(decoded)
```

## Контрольні питання

### 1. Що таке жадібні алгоритми?

Жадібні алгоритми — це алгоритми, які на кожному кроці обирають локально оптимальне рішення. Вони не повертаються до попередніх кроків і намагаються отримати глобально оптимальний результат.

### 2. Що таке префіксний код? Який код використовується у коді Гафмена?

Префіксний код — це код, у якому жоден код символу не є початком іншого коду. У алгоритмі Гафмена використовується двійковий префіксний код.

### 3. Як пов’язана купа з чергою з пріоритетами?

Купа є структурою даних, яка використовується для реалізації черги з пріоритетами. Вона дозволяє швидко знаходити та видаляти елемент із найменшим або найбільшим пріоритетом.

### 4. Що таке стиснення даних?

Стиснення даних — це процес зменшення обсягу даних для економії пам’яті та пришвидшення передачі інформації.

### 5. Основні кроки алгоритму Гафмена

1. Підрахунок частот символів.
2. Створення вузлів дерева.
3. Побудова мінімальної купи.
4. Об’єднання вузлів із найменшими частотами.
5. Побудова дерева Гафмена.
6. Генерація кодів символів.

### 6. Недоліки алгоритму Гафмена

- Потрібно зберігати дерево кодів.
- Не завжди ефективний для малих повідомлень.
- Поступається сучасним алгоритмам стиснення.
