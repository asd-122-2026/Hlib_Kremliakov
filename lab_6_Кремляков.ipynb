## Виконання

```python
class Stack:

    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def size(self):
        return len(self.items)
```

### Перевірка стеку

```python
s = Stack()

s.push("hello")
s.push("true")

print("Розмір стеку:", s.size())

print("Верхній елемент:", s.peek())

print("Видалений елемент:", s.pop())

print("Стек після видалення:", s.items)
```

### Самостійне завдання — pop_n()

```python
def pop_n(stack, n):

    if n >= stack.size():
        print("Помилка: n більше за розмір стеку")
        return

    for i in range(n + 1):
        removed = stack.pop()
        print("Видалено:", removed)
```

### ПЕРЕВІРКА

```python
s = Stack()

s.push(1)
s.push(2)
s.push(3)
s.push(4)
s.push(5)

print("Початковий стек:", s.items)

pop_n(s, 2)

print("Стек після видалення:", s.items)
```

## Асимптотична складність операцій стеку

### push()
Складність: 

Додавання елемента у стек виконується за константний час.

### pop()
Складність: 

Видалення верхнього елемента також виконується за константний час.

### search()
Складність: 

Для пошуку елемента необхідно переглянути стек.

### Реалізація черги

```python
class Queue:

    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def enqueue(self, item):
        self.items.insert(0, item)

    def dequeue(self):
        return self.items.pop()

    def size(self):
        return len(self.items)
```

### Перевірка черги

```python
q = Queue()

q.enqueue(1)
q.enqueue(2)
q.enqueue(3)

print("Черга:", q.items)

print("Видалений елемент:", q.dequeue())

print("Черга після видалення:", q.items)
```

### Самостійне завдання — print_n()

```python
def print_n(queue, n):

    if n >= queue.size():
        print("Помилка: n більше за розмір черги")
        return

    print("Елементи черги:")

    for i in range(n + 1):
        print(queue.items[-(i + 1)])
```

### ПЕРЕВІРКА

```python
q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)

print("Черга:", q.items)

print_n(q, 2)
```

## Асимптотична складність операцій черги

### enqueue()
Складність: 

Операція insert(0, item) потребує зміщення елементів.

### dequeue()
Складність: 

Видалення останнього елемента виконується швидко.

### search()
Складність: 

Пошук потребує перегляду елементів черги.

## Контрольні питання

### 1. Що таке стек і які операції можна виконувати зі стеком?

Стек — це структура даних, яка працює за принципом LIFO (Last In First Out), тобто останній доданий елемент видаляється першим.

Основні операції стеку:
- push() — додавання елемента;
- pop() — видалення верхнього елемента;
- peek() — перегляд верхнього елемента;
- isEmpty() — перевірка на порожність;
- size() — визначення кількості елементів.


### 2. Яка основна відмінність між стеком та чергою?

Стек працює за принципом LIFO — останній прийшов, перший вийшов.

Черга працює за принципом FIFO — перший прийшов, перший вийшов.

У стеку додавання і видалення відбувається з одного кінця, а в черзі — з різних.


### 3. Як ви можете реалізувати стек за допомогою масиву і за допомогою зв’язаного списку? Які переваги та недоліки кожного підходу?

Стек можна реалізувати:
- за допомогою масиву (списку Python);
- за допомогою зв’язаного списку.

Переваги масиву:
- проста реалізація;
- швидкий доступ до елементів.

Недоліки:
- можливе перевиділення пам’яті.

Переваги зв’язаного списку:
- динамічний розмір;
- ефективне додавання елементів.

Недоліки:
- складніша реалізація;
- більші витрати пам’яті.


### 4. Які є застосування стека та черги в програмуванні і реальному житті?

Застосування стеку:
- історія браузера;
- виклики функцій;
- перевірка дужок;
- алгоритми рекурсії.

Застосування черги:
- друк документів;
- обробка задач процесором;
- системи обслуговування;
- черги повідомлень.
